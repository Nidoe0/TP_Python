# TP 3 — Le stock d'une épicerie (dictionnaires)

```python
import json
from collections import Counter
from copy import deepcopy

stock = {
    "riz": {"prix": 3200, "quantite": 50},
    "huile": {"prix": 9500, "quantite": 12},
    "sucre": {"prix": 4000, "quantite": 30},
    "savon": {"prix": 1500, "quantite": 0},
}

# Q1
for produit, info in stock.items():
    print(f"{produit} : {info['prix']} Ar ; quantité = {info['quantite']}")

# Q2
valeur_totale = sum(info["prix"] * info["quantite"] for info in stock.values())
print("Valeur totale :", f"{valeur_totale:,}".replace(",", " "), "Ar")

# Q3
rupture = [nom for nom, info in stock.items() if info["quantite"] == 0]
print("Rupture :", rupture)

# Q4
def vendre(stock, produit, quantite):
    if quantite <= 0:
        return False, "La quantité doit être positive."
    article = stock.get(produit)
    if article is None:
        return False, f"Produit inconnu : {produit}."
    if article["quantite"] < quantite:
        return False, f"Stock insuffisant pour {produit}."
    article["quantite"] -= quantite
    return True, f"Vente réussie : {quantite} {produit}."

stock_test = deepcopy(stock)
print(vendre(stock_test, "riz", 3))
print(vendre(stock_test, "inconnu", 1))
print(vendre(stock_test, "savon", 1))

# Q5
def ajouter_produit(stock, nom, prix, quantite):
    if nom in stock:
        return False, f"{nom} existe déjà."
    if prix < 0 or quantite < 0:
        return False, "Prix et quantité doivent être positifs ou nuls."
    stock[nom] = {"prix": prix, "quantite": quantite}
    return True, f"{nom} ajouté."

print(ajouter_produit(stock_test, "lait", 5000, 8))
print(ajouter_produit(stock_test, "riz", 3300, 10))

# Q6
prix_hausse = {nom: round(info["prix"] * 1.10) for nom, info in stock.items()}
print("Prix +10 % :", prix_hausse)

# Q7
ventes = ["riz", "sucre", "riz", "huile", "riz"]
compteur = {}
for produit in ventes:
    compteur[produit] = compteur.get(produit, 0) + 1
print("Avec get :", compteur)
print("Avec Counter :", dict(Counter(ventes)))

# Q8 — Le fichier sera enregistré dans le répertoire courant de Jupyter
with open("stock.json", "w", encoding="utf-8") as fichier:
    json.dump(stock, fichier, ensure_ascii=False, indent=2)
with open("stock.json", "r", encoding="utf-8") as fichier:
    stock_recharge = json.load(fichier)
print("Stock identique après rechargement :", stock == stock_recharge)
```
