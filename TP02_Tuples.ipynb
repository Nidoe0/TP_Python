# TP 2 — Les villes de Madagascar (tuples)

```python
from math import dist, radians, sin, cos, asin, sqrt
from itertools import combinations
from collections import namedtuple

villes = [
    ("Antananarivo", -18.88, 47.51),
    ("Toamasina", -18.15, 49.40),
    ("Mahajanga", -15.72, 46.32),
    ("Fianarantsoa", -21.45, 47.09),
]

# Q1 — Déballage
for nom, lat, lon in villes:
    print(f"{nom} : {lat}, {lon}")

# Q2 — La latitude la plus grande correspond à la ville la plus au nord
ville_nord = max(villes, key=lambda v: v[1])
print("Ville la plus au nord :", ville_nord[0])

# Q3 — Bornes de latitude et longitude
def bornes(liste_villes):
    latitudes = [v[1] for v in liste_villes]
    longitudes = [v[2] for v in liste_villes]
    return min(latitudes), max(latitudes), min(longitudes), max(longitudes)

lat_min, lat_max, lon_min, lon_max = bornes(villes)
print("Bornes :", lat_min, lat_max, lon_min, lon_max)

# Q4 — Distance euclidienne en degrés
def distance(p1, p2):
    return dist(p1, p2)

# Q5 — Toutes les paires de villes
distances = {}
for v1, v2 in combinations(villes, 2):
    distances[(v1[0], v2[0])] = distance((v1[1], v1[2]), (v2[1], v2[2]))
print("Nombre de paires :", len(distances))
for paire, d in distances.items():
    print(paire, ":", round(d, 4), "degrés")

# Q6 — Les tuples sont immuables
try:
    villes[0][1] = -18.90
except TypeError as erreur:
    print("Modification impossible :", erreur)

# Correction : créer un nouveau tuple
ville_corrigee = (villes[0][0], -18.90, villes[0][2])
villes_corrigees = [ville_corrigee] + villes[1:]
print("Nouvelle ville :", villes_corrigees[0])

# Q7 Bonus — namedtuple et haversine
Ville = namedtuple("Ville", ["nom", "latitude", "longitude"])
villes_nommees = [Ville(*v) for v in villes]

def distance_km(v1, v2):
    lat1, lon1 = radians(v1.latitude), radians(v1.longitude)
    lat2, lon2 = radians(v2.latitude), radians(v2.longitude)
    dlat, dlon = lat2 - lat1, lon2 - lon1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    return 2 * 6371 * asin(sqrt(a))

print("Tana-Toamasina :", round(distance_km(villes_nommees[0], villes_nommees[1]), 2), "km")
```
