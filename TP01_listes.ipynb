# TP 1 — Les notes d'une classe (listes)

```python
notes = [12, 8, 15, 17, 9, 11, 14, 6, 13, 16]

# Q1 — Statistiques
print("Nombre :", len(notes))
print(f"Moyenne : {sum(notes) / len(notes):.2f}")
print("Minimum :", min(notes), "| Maximum :", max(notes))

# Q2 — Notes admises et ajournés
notes_admises = [n for n in notes if n >= 10]
ajournes = [n for n in notes if n < 10]
print("Notes >= 10 :", notes_admises)
print("Nombre d'ajournés :", len(ajournes))

# Q3 — sorted crée une nouvelle liste
print("Tri croissant :", sorted(notes))
print("Trois meilleures :", sorted(notes, reverse=True)[:3])
print("Original conservé :", notes)

# Q4 — Bonus de 2, plafonné à 10 pour les notes < 10
notes_deliberation = [min(n + 2, 10) if n < 10 else n for n in notes]
print("Après délibération :", notes_deliberation)

# Q5 — Histogramme
def histogramme(liste_notes):
    for mini, maxi in [(0, 4), (5, 9), (10, 14), (15, 20)]:
        nombre = sum(1 for n in liste_notes if mini <= n <= maxi)
        print(f"{mini:>2}-{maxi:<2} | {'*' * nombre}")

histogramme(notes)

# Q6 — Rang dans la liste triée par ordre décroissant
def rang(liste_notes, note):
    classement = sorted(liste_notes, reverse=True)
    return classement.index(note) + 1 if note in classement else None

print("Rang de 14 :", rang(notes, 14))

# Q7 — Fusion sans doublons en gardant la première apparition
fusion = []
for n in notes + [15, 6, 18]:
    if n not in fusion:
        fusion.append(n)
print("Fusion :", fusion)
```
